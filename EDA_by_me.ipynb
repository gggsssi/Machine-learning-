```python
with zipfile.ZipFile(zip_path,"r") as z:
    files=z.extract("profile.txt")
ps1_p=pd.read_csv("profile.txt",sep="\t",header=None)
ps1_p.columns = [
    "cooler",
    "valve",
    "pump_leakage",
    "accumulator",
    "stability"
]
cycles = [2, 34, 56, 78, 200, 270, 300, 500, 550, 570, 620]

plt.figure(figsize=(12, 5))

for i in cycles:
    leakage = ps1_p.loc[i, "pump_leakage"]
    plt.plot(
        time_ps1[2000:5001],
        ps1.iloc[i, 2000:5001],
         label=f"Cycle {i} | Leakage={leakage}"
    )
    

plt.xlabel("Time (seconds)")
plt.ylabel("Pressure (bar)")
plt.title("PS1 pressure: 20–50 seconds")
plt.legend()

plt.show()
```

```python
cycles = [2, 34, 56, 78, 200, 270, 300, 500, 550, 570, 620]

plt.figure(figsize=(12, 5))

for i in cycles:
    press = ps1_p.loc[i, "accumulator"]
    plt.plot(
        time_ps1[2000:5001],
        ps1.iloc[i, 2000:5001],
         label=f"Cycle {i} | Press={press}"
    )
    

plt.xlabel("Time (seconds)")
plt.ylabel("Pressure (bar)")
plt.title("PS1 pressure: 20–50 seconds")
plt.legend()

plt.show()
```
